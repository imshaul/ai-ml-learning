# **GroupBy**

```python
df.groupby('Segment').count()
```

```python
df.groupby('Segment').max()
```

```python
df.groupby('Sales').agg({'Quantity':['sum','min','max'],'Ship Mode':['sum','min','max']})
```

```python
df.groupby(['Sales','Profit']).agg({'Quantity':['sum','min','max']})
```

```python
df['Profit'].describe()
```

```python
total_op = df.groupby('Category')
total_op['Sales'].sum()
```

```python
total_op['Profit'].sum()
```

```python
total_op['Sales'].mean()
```

```python
total_op['Quantity'].count()
```

```python
total_sales = df.groupby('Country')
total_sales['Sales'].sum().sort_values(ascending = False).head(10)
```

```python
category_op = df.groupby('Category')
category_op[['Sales','Profit','Quantity']].agg({'Sales':['sum','mean'],
                                                'Profit':['sum','mean'],
                                               'Quantity':['sum','mean']})
```

```python
category_op['Quantity'].agg()
```

```python
region_op = df.groupby('Region')
region_op[['Sales','Profit']].agg({'Sales':['sum']
                                   ,'Profit':['sum','mean']})
```

```python
df.columns
```

```python
# for segment - total sales,profit and avg sales,profit
# for category + subcate combo calculate sales and profit
combined_category = df.groupby(['Category','Sub-Category'])
combined_category[['Sales','Profit']].agg('sum')
```

```python
total_seg = df.groupby('Segment')
total_seg[['Sales','Profit']].agg({'Sales':'mean',
                                  'Profit':'mean'})
```
