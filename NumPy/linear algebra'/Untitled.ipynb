```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([5,4,1])
b = np.array([4,5,2])
c = np.array([1,2,5])
```

```python
dot_product_ab = np.dot(a,b)
cos_theta_ab = dot_product_ab/(np.linalg.norm(a)*np.linalg.norm(b))
angle_ab = np.degrees(np.arccos(cos_theta_ab))
print(f"Angle between A and B : {angle_ab:.0f} degree")
```

```python
ax = plt.figure(figsize=(6,6)).add_subplot(projection = '3d')
ax.quiver(0,0,0,*a,color = "r",linewidth=3)
ax.quiver(0,0,0,*b,color = "r",linewidth=3)
ax.text(*a,"A");ax.text(*b,"B")
ax.set_title(f"Angle = {angle_ab:.2f}")
plt.show()
```
